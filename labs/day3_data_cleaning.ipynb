```python
## ⚙️ Step 1: Import Pandas & Read  Titanic  Data From Datasets
import pandas as pd
df=pd.read_csv(r'C:\Users\ASHISH RAJ\Desktop\ai-journey\datasets\Titanic-Dataset.csv')
df.head(5)
```

## Handle Missing Values

```python
print(df.isnull().sum())
```

```python
# Option 1: Drop rows with any missing values
df_droped=df.dropna()
df_droped.shape
```

```python
# Option 2: Fill missing age through mean value
df_filled=df.fillna({'Age':df['Age'].mean()})
df_filled.head(5)
```

```python
df_filled['Age'].isnull().sum()
```

```python
# Check current data types
df['Age'].dtype
```

```python
df_filled["Age"] = df_filled["Age"].astype("int")
```

```python
# Check for duplicate rows
df.duplicated().sum()
```

```python
# Standardize the 'Sex' column
df['Sex']=df['Sex'].str.lower().str.strip()
df['Sex'].head(5)
```

```python

# Calculate IQR for 'Fare'
Q1=df['Fare'].quantile(0.25)
Q3=df['Fare'].quantile(0.75)
IQR=Q3-Q1
IQR
```

```python
# Define bounds
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
```

```python
# Filter out outliers
df_clean=df[(df['Fare']>=lower_bound)&(df['Fare']<=upper_bound)]
df_clean
```
